```python
#importing the dataset
import pandas as pd 
import matplotlib.pyplot as plt
data = pd.read_csv(r"C:\Users\Dell\Downloads\Titanic-Dataset (1).csv")
data
```

```python
#bar graph from mat plotlib
counts = data["Pclass"].value_counts().sort_index()
plt.bar(counts.index,counts.values)
plt.xlabel("Passenger class")
plt.ylabel("Number of Passanger")
plt.title("Number of PaSSANGER in each class")
plt.xticks([1,2,3])
```

```python
#Bar graph in a specific clasification
counts = data.groupby(["Survived","Sex"]).size().unstack()
counts.plot(kind="bar")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passangers")
plt.title("Titanic Survival by Gender")
plt.xticks([0,1],[" Not Survived","Survived"])
```

```python
#Scatter plot of a data set
data = data.dropna(subset = ["Age","Fare"])
#Create scatter plot 
plt.scatter(data["Age"] , data["Fare"])
plt.xlabel("age")
plt.ylabel("Fare")
plt.title("Age vs Fare in Titnic Dataset")
```
