```python
#Creating List Of Students
students = ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Neha"]

#Creating Tuple Of Students Roll Number
roll_numbers = (20,21,22,23,24,25)

#Creating Dictionary Of Student Data
student_data = {
    "Name":"Rahul",
    "Age" : 18,
    "Corse" : "Cse"
       }
print(students)
print(roll_numbers)
print(student_data)
```

```python
#performing operations on list add delete and update
students = ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Neha"]
#In list we can add element using append() and insert method()
#append() adds the element at the end of the list
students.append("omkar")
print(students)
#insert adds the element at a specific position
students.insert(2,"Aman")
print(students)
```

```python
#updating element in the list
students[0] = "Sharma"
print(students)
```

```python
#deleting elements from the list
students.remove("Priya")
print(students)
```

```python
# Perform operations add delete update on dictionary
student_data = {
    "Name":"Rahul",
    "Age" : 18,
    "Corse" : "Cse"
       }
#Adding key value pair in dictionary
student_data["Joining data"] = 2026
print(student_data)
```

```python
#updating key value pair
student_data["Name"] = "Omkar"
print(student_data)
```

```python
#Deleting values in dictionary
del student_data["Corse"] 
print(student_data)
```
