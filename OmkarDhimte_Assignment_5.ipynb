```python
#To check weather the string contains a-z A-Z 0-9 or other character using regular expression(re)
import re 
string = input("Enter the string")
if re.fullmatch(r'[a-zA-Z0-9]+' , string):
    print("The string contains a-z A-Z and 0-9")
else:
    print("The string contain only characters")
```
